# UPI India Transaction Analytics

This notebook performs an exploratory analysis of UPI transaction data in India. It covers loading the dataset, handling missing values, and conducting various analyses such as calculating transaction success rates, average transaction amounts by gender and UPI app, and identifying top cities by transaction volume.

## Load Data

Loading the `UPI_India_Transaction_Analytics_100k.xlsx` file into a pandas DataFrame.

```python
# Loads the transaction data from an Excel file into a pandas DataFrame and displays the first few rows.
import pandas as pd

df = pd.read_excel('/content/UPI_India_Transaction_Analytics_5000.xlsx')
display(df.head())
```

```python
# Displays the first few rows of the DataFrame `df`.
df.head()
```

```python
# Displays the column names of the DataFrame `df`.
df.columns
```

```python
# Fills missing values in 'Gender' and 'Employment_Status' columns with 'Unknown'.
df['Gender'] = df['Gender'].fillna('Unknown')
df['Employment_Status'] = df['Employment_Status'].fillna('Unknown')
```

```python
# Fills missing values in 'Monthly_Income_Band' with the mode of the column.
df['Monthly_Income_Band'] = df['Monthly_Income_Band'].fillna(df['Monthly_Income_Band'].mode()[0])
```

```python
# Fills missing values in 'Network_Type' column with 'Unknown'.
df['Network_Type'] = df['Network_Type'].fillna('Unknown')
```

```python
# Fills missing values in 'Merchant_Category' column with 'Unknown'.
df['Merchant_Category'] = df['Merchant_Category'].fillna('Unknown')
```
